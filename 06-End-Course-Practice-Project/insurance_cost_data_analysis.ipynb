# Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import cross_val_score, train_test_split
```

# Downloading the dataset

```python
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/medical_insurance_dataset.csv'

df = pd.read_csv(url, header=None)
df.to_csv("Insurance Cost Analysis Dataset.csv", index= False)
df
```

# Import the dataset

```python
print(df.head(10))
```

```python
df.columns =  ["age", "gender", "bmi", "no_of_children", "smoker", "region", "charges"]
df.to_csv("Insurance Cost Analysis Dataset.csv", index = False)
df
```

```python
df.replace("?", np.nan, inplace = True)
df
```

# Data Wrangling

```python
print(df.info())
```

Handling missing values


```python
mean_age = df["age"].astype("float").mean(axis=0)
df["age"].replace(np.nan, mean_age, inplace=True)

is_smoker = df["smoker"].value_counts().idxmax()
df["smoker"].replace(np.nan, is_smoker, inplace = True)

df[["age","smoker"]] = df[["age","smoker"]].astype("int")
print(df.info())
```

```python
df[["charges"]] = np.round(df[["charges"]],2)
print(df.head())
```

# Exploratory Data Analysis (EDA)
